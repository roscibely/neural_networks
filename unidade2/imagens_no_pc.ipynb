# Representação e Manipulação de Imagens no Computador

Antes de começarmos a trabalhar com Redes Neurais Convolucionais (CNNs), precisamos entender como os computadores enxergam e manipulam imagens.

Uma imagem digital é, essencialmente, uma grade de números (uma matriz). Cada elemento dessa matriz é chamado de **pixel** (picture element) e representa a intensidade de cor daquela posição.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Imagens em Tons de Cinza (2D)

Em uma imagem em tons de cinza, cada pixel é representado por um único valor numérico, geralmente variando de 0 (preto absoluto) a 255 (branco absoluto).

Vamos criar uma imagem artificial de 5x5 pixels para visualizar esses valores numéricos de forma explícita.

```python
matriz_cinza = np.array([
    [0, 50, 100, 150, 200],
    [50, 100, 150, 200, 255],
    [100, 150, 200, 255, 150],
    [150, 200, 255, 150, 100],
    [200, 255, 150, 100, 50]
])

print("Matriz de Pixels:")
print(matriz_cinza)
print("\nDimensões da Imagem:", matriz_cinza.shape)
```

Agora, vamos converter essa matriz numérica em uma imagem visualizável utilizando o matplotlib.

```python
plt.imshow(matriz_cinza, cmap='gray', vmin=0, vmax=255)
plt.colorbar()
plt.show()
```

## 2. Imagens Coloridas (3D / Canais RGB)

Imagens coloridas geralmente utilizam o sistema RGB (Red, Green, Blue). Em vez de uma única matriz, temos três matrizes empilhadas (uma para cada canal de cor).

Vamos construir uma imagem colorida de 3x3 pixels onde cada pixel contém 3 valores correspondentes a [Vermelho, Verde, Azul].

```python
imagem_rgb = np.zeros((3, 3, 3), dtype=np.uint8)

imagem_rgb[0, 0] = [255, 0, 0]
imagem_rgb[0, 1] = [0, 255, 0]
imagem_rgb[0, 2] = [0, 0, 255]

imagem_rgb[1, 0] = [255, 255, 0]
imagem_rgb[1, 1] = [0, 255, 255]
imagem_rgb[1, 2] = [255, 0, 255]

imagem_rgb[2, 0] = [255, 255, 255]
imagem_rgb[2, 1] = [128, 128, 128]
imagem_rgb[2, 2] = [0, 0, 0]

print("Dimensões da Imagem RGB:", imagem_rgb.shape)
```

Visualizando a imagem colorida criada:

```python
plt.imshow(imagem_rgb)
plt.show()
```

## 3. Isolando Canais de Cor

Podemos fatiar nossa matriz tridimensional para analisar cada canal de cor separadamente. Isso nos ajuda a entender como os filtros funcionam em uma rede convolucional.

```python
canal_vermelho = imagem_rgb[:, :, 0]
canal_verde = imagem_rgb[:, :, 1]
canal_azul = imagem_rgb[:, :, 2]

fig, eixos = plt.subplots(1, 3, figsize=(10, 3))

eixos[0].imshow(canal_vermelho, cmap='Reds', vmin=0, vmax=255)
eixos[0].set_title('Canal Vermelho (R)')

eixos[1].imshow(canal_verde, cmap='Greens', vmin=0, vmax=255)
eixos[1].set_title('Canal Verde (G)')

eixos[2].imshow(canal_azul, cmap='Blues', vmin=0, vmax=255)
eixos[2].set_title('Canal Azul (B)')

plt.show()
```

## Conexão com CNNs

Nas Redes Neurais Convolucionais, os neurônios das primeiras camadas não olham para a imagem inteira de uma vez. Eles aplicam pequenas matrizes de pesos (chamadas de filtros ou kernels) que deslizam sobre essas matrizes numéricas que acabamos de estudar, realizando operações matemáticas para extrair características visuais como bordas, texturas e formas.
